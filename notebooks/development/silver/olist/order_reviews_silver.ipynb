# Order Reviews - Silver Transformation

## Setup

```python
from pyspark.sql.functions import col, trim

environment = "dev"

catalog = f"ecommerce_{environment}"
source_table = f"{catalog}.bronze.olist_order_reviews"
target_table = f"{catalog}.silver.olist_order_reviews"
```

## Read Bronze data

```python
bronze_df = spark.table(source_table)
```

## Review the data

```python
bronze_df.printSchema()
```

```python
display(bronze_df.limit(10))
```

```python
print("Number of rows:", bronze_df.count())
print("Number of columns:", len(bronze_df.columns))
```

```python
rescued_row_count = (
    bronze_df
    .filter(col("_rescued_data").isNotNull())
    .filter(trim(col("_rescued_data")) != "")
    .count()
)

print("Number of rescued rows:", rescued_row_count)
```

```python
for column, dtype in bronze_df.dtypes[:7]:
    print(column)
    print("Null count:", bronze_df.filter(col(column).isNull()).count())
    print("Distinct count:", bronze_df.filter(col(column).isNotNull()).select(column).distinct().count())

    if dtype == "string":
        print("Extra whitespace row count:",
            (
                bronze_df.withColumn(f"{column}_trimmed", trim(col(column)))
                .filter(col(column) != col(f"{column}_trimmed"))
                .count()
            )
        )
    print("-"*20)
```

```python
bronze_df.count() == bronze_df.select("review_id","order_id").distinct().count()
```

The combination of review_id and order_id is unique and identifies each review record.

## Transform to Silver

```python
silver_df = (
    bronze_df
    .withColumn(
        "review_comment_title",
        trim(col("review_comment_title"))
    )
    .withColumn(
        "review_comment_message",
        trim(col("review_comment_message"))
    )
)
```

```python
print(
    "Empty review titles after trimming:",
    silver_df.filter(col("review_comment_title") == "").count()
)

print(
    "Empty review messages after trimming:",
    silver_df.filter(col("review_comment_message") == "").count()
)
```

```python
display(
    silver_df.filter(
        (col("review_comment_title") == "") | (col("review_comment_message") == "")
    )
    .select("review_id", "order_id", "review_comment_title", "review_comment_message")
)
```

```python
silver_df = silver_df.replace(
    "",
    None,
    subset=["review_comment_title", "review_comment_message"]
)
```

```python
print(
    "Empty review titles:",
    silver_df.filter(col("review_comment_title") == "").count()
)

print(
    "Empty review messages:",
    silver_df.filter(col("review_comment_message") == "").count()
)
```

## Write to Silver

```python
(
    silver_df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(target_table)
)
```

## Review the result

```python
silver_table_df = spark.table(target_table)

silver_table_df.printSchema()
```

```python
display(silver_table_df.limit(10))
```
