# Category Tree - Silver Transformation

## Parameters

```python
dbutils.widgets.text(
    name="environment",
    defaultValue="dev",
    label="Environment"
)

environment = dbutils.widgets.get("environment").strip().lower()

if environment not in ("dev", "test", "prod"):
    raise ValueError(
        f"Unsupported environment: {environment}. Expected dev, test, or prod."
    )
```

## Setup

```python
catalog = f"ecommerce_{environment}"
source_table = f"{catalog}.bronze.retailrocket_category_tree"
target_table = f"{catalog}.silver.retailrocket_category_tree"
```

## Read Bronze data

```python
bronze_df = spark.table(source_table)
```

## Transform to Silver

```python
silver_df = (
    bronze_df
    .withColumnRenamed("categoryid", "category_id")
    .withColumnRenamed("parentid", "parent_category_id")
)
```

## Write to Silver

```python
(
    silver_df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(target_table)
)
```
